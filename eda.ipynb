```python
import numpy as np

# Stack the per-frame 6-vectors into (T, 6) arrays
obs = np.vstack(df['observation.right_hand_q'].to_numpy())      # (T, 6)
teleop = np.vstack(df['teleop.right_hand_joints'].to_numpy())   # (T, 6)
t = df['timestamp'].to_numpy()

fig, axes = plt.subplots(3, 2, figsize=(14, 10), sharex=True)
for j, ax in enumerate(axes.ravel()):
    ax.plot(t, obs[:, j], label='observation.right_hand_q', color='tab:blue')
    ax.plot(t, teleop[:, j], label='teleop.right_hand_joints', color='tab:orange', linestyle='--')
    ax.set_title(f'Joint {j}')
    ax.set_ylabel('value')
    ax.grid(True, alpha=0.3)
axes[-1, 0].set_xlabel('time (s)')
axes[-1, 1].set_xlabel('time (s)')
axes[0, 0].legend(loc='best')
fig.suptitle('Right hand: observation vs teleop (6 joints)')
fig.tight_layout()
plt.show()
```
